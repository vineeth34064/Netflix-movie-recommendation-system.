```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

importing dataset

```python
dataset = pd.read_csv('netflix.csv')
```

```python
dataset
```

```python
transactions = dataset.iloc[:7466, :20].astype(str).values.tolist()

```

```python
transactions #this will result in getting 2d array
```

```python
from apyori import apriori
netflix = apriori(transactions=transactions, min_support = 0.003,min_confidence=0.2,min_lift = 3,min_len = 2)
```

```python
netflix
```

#Visualizing

```python
results = list(netflix)
```

```python
results
```

#displaying results in proper way

```python
def inspect(results):
  lhs = [tuple(result[2][0][0])[0] for result in results]
  rhs = [tuple(result[2][0][1])[0] for result in results]

  supports = [result[1] for result in results]
  return list(zip(lhs,rhs,supports))
```

```python
resultdata = pd.DataFrame(inspect(results),columns=['movie1','movie2','support'])
```

```python
resultdata
```

```python
resultdata.nlargest(n = 30,columns='support')
```
